**Step 2:** Create a GPT object.

**Step 3:** Train it for some number of iterations on a text of your choice. <br>Provide the path to the text.
You can alternatively provide it plain text.
**Other options:**
* plot: True by default. Plot the loss function.
* update: Default 50. How many iterations to be updated on the progress. Use 0 for no updates.
* sample_prompt: Test prompt to compare model response before and after training. By default, no querying will be done.

If you don't want to wait for the model to train, try setting the iterations to 100, or skip a few steps to load in the pre-trained model
If ploting, observe the loss plot after the model is trained.

There are two sample training examples below. One is for the pre-downloaded recipes text, the other will get a state of the union dataset from Kaggle and download it. Uncomment the one you want to use (or use both if you really want to)

```python
path = Vocab.get_state_of_union_file()
my_gpt.train(n_iter = 1000, text_file=path, sample_prompt = "Tell me about the economy, specifically unemployment, inflation, tariffs and taxes.")

# If having issues with kaggle, or want a different dataset, uncomment this line to use the pre-downloaded dataset of recipes
# my_gpt.train(n_iter = 1000, text_file='./texts/recipes.txt', sample_prompt = "make a salad with tomatoes, olives and walnuts")
```

**Step 4:** Before doing anything else, save the model.
It saves the model parameters as well as the vocabulary the model is associated with.
Just enter the name of the model, not the file path

```python
my_gpt.save('state_of_union_model')
```

**Step 5:** If starting fresh, you can load the model like so

```python
my_gpt = GPT.load_gpt_from('state_of_union_model')
```

**Step 6:** Now, query the model!

```python
my_gpt.query("Tell me about the economy, specifically unemployment, inflation, tariffs and taxes.")
```
